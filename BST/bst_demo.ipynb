```python
from bst import BinarySearchTree
```

```python
tree = BinarySearchTree()
```

```python
vals = [2,1,8,4,26,41,7,89,9,23,15,55,98,74,11,44,54,18,77,42,20,33,66,23,24,25] # Agregar 25 valores aleatorios que permitan demostrar los 3 casos de delete

for val in vals:
    tree.insert(val)
```

```python
tree.print_pretty()
```

## Delete - caso no. 1

Caso donde se elimina una leaf 

```python
tree.delete(20)
tree.print_pretty()
```

## Delete - caso no. 2

Se elimina un nodo con solo 1 hijo

```python
tree.delete(24)
tree.print_pretty()
```

## Delete - caso no. 3

Con sucesor in order  se elimina un nodo con 2 hijos

```python
tree.delete(15)
tree.print_pretty()
```
